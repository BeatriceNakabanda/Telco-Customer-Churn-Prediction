```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10,6)

# Load processed data
df = pd.read_csv("../data/processed/telco_cleaned.csv")
df.head()
```

```python
# Correlation with target
correlations = df.corrwith(df['Churn']).sort_values(ascending=False)
top_corr = correlations[1:11]  # top 10 excluding target
print("Top correlated features with Churn:\n", top_corr)

# Plot correlations
top_corr.plot(kind='barh')
plt.title("Top 10 Features Correlated with Churn")
plt.xlabel("Correlation with Churn")
plt.savefig("../outputs/figures/eda_top_correlations.png", bbox_inches='tight')
plt.show()
```

```python
# Example: Tenure and Monthly Charges distribution by churn
sns.histplot(data=df, x='Tenure in Months', hue='Churn', multiple='stack', bins=30)
plt.title("Tenure Distribution by Churn")
plt.xlabel("Tenure (Months)")
plt.ylabel("Number of Customers")
plt.savefig("../outputs/figures/eda_tenure_churn.png", bbox_inches='tight')
plt.show()

sns.histplot(data=df, x='Monthly Charge', hue='Churn', multiple='stack', bins=30)
plt.title("Monthly Charge Distribution by Churn")
plt.xlabel("Monthly Charge")
plt.ylabel("Number of Customers")
plt.savefig("../outputs/figures/eda_monthly_charge.png", bbox_inches='tight')
plt.show()
```

```python
# Example: Contract type vs Churn
sns.countplot(data=df, x='Contract_Month-to-month', hue='Churn')
plt.title("Churn by Month-to-Month Contract")
plt.xlabel("Month-to-Month Contract (1=Yes,0=No)")
plt.ylabel("Number of Customers")
plt.savefig("../outputs/figures/eda_contract_churn.png", bbox_inches='tight')
plt.show()
```

```python
# Load churn risk if available
try:
    churn_df = pd.read_csv("../outputs/tables/high_risk_customers.csv")
    sns.histplot(churn_df['Churn_Risk'], bins=30)
    plt.title("Distribution of Churn Risk for Customers")
    plt.xlabel("Churn Probability")
    plt.ylabel("Number of Customers")
    plt.savefig("../outputs/figures/eda_churn_risk.png", bbox_inches='tight')
    plt.show()
except FileNotFoundError:
    print("High risk customers file not found. Run run_pipeline.py first.")
```
